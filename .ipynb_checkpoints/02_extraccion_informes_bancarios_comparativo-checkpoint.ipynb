# Extracción documental de informes bancarios
El notebook usa `datos/Banco/indice.csv`, procesa los PDF originales y conserva la página y las coordenadas de cada hallazgo. Empieza con un informe por clasificadora. En la sección de ejecución puedes cambiar a la modalidad de lote completo.

La secci?n final compara motores con referencias anotadas y exporta las m?tricas.

## 1. Configurar el corpus y el OCR
Las coordenadas están en puntos PDF, con origen arriba a la izquierda en la página sin rotar. `regiones_tabla_ocr` admite regiones de tablas escaneadas seleccionadas en la vista previa. Usa como clave la ruta relativa con barras `/` y la página empezando en 1.

```python
import csv
import hashlib
import json
import os
import re
import shutil
from pathlib import Path

import pandas as pd
import pymupdf
import pytesseract
from IPython.display import Image, display
from PIL import Image as ImagenPIL

raiz = Path.cwd()
corpus = (raiz / "datos" / "Banco").resolve()
indice = corpus / "indice.csv"
salida = raiz / "salidas" / "banco_comparativo"
dpi_ocr = 250
area_minima_para_ocr = 0.08
guardar_imagenes = True
regiones_tabla_ocr = {}
# Ejemplo: regiones_tabla_ocr[("BANCO/CLASIFICADORA/informe.pdf", 4)] = [(50, 100, 550, 350)]

tesseract = os.getenv("TESSERACT_CMD") or shutil.which("tesseract")
if not tesseract:
    tesseract_local = Path(os.getenv("ProgramFiles", r"C:\Program Files")) / "Tesseract-OCR" / "tesseract.exe"
    if tesseract_local.is_file():
        tesseract = str(tesseract_local)
if tesseract:
    pytesseract.pytesseract.tesseract_cmd = tesseract
if "spa" not in pytesseract.get_languages():
    raise RuntimeError("Tesseract necesita el idioma spa para ejecutar el OCR.")
print("Tesseract:", pytesseract.get_tesseract_version())
```

## 2. Validar el inventario
Se verifican las rutas del índice, la extensión de los archivos y que las rutas queden dentro del corpus. La huella SHA-256 se calcula al procesar cada PDF. Los metadatos del índice mantienen su propio origen.

```python
with indice.open(encoding="utf-8-sig", newline="") as archivo:
    lector = csv.DictReader(archivo)
    requeridas = {"archivo", "estado", "entidad", "clasificadora", "periodo", "clasificacion", "url"}
    faltantes = requeridas - set(lector.fieldnames or [])
    if faltantes:
        raise ValueError(f"Columnas faltantes en {indice}: {sorted(faltantes)}")
    catalogo = list(lector)

inventario = {}
incidencias_indice = []
for fila in catalogo:
    referencia = fila["archivo"] or ""
    if fila["estado"] != "descargado":
        incidencias_indice.append({"archivo": referencia, "motivo": fila["estado"], "detalle": fila.get("error")})
        continue
    ruta = (corpus / referencia.replace("\\", "/")).resolve()
    if not ruta.is_relative_to(corpus) or ruta.suffix.lower() != ".pdf" or not ruta.is_file():
        incidencias_indice.append({"archivo": referencia, "motivo": "ruta PDF inválida", "detalle": ""})
        continue
    relativa = ruta.relative_to(corpus).as_posix()
    if relativa in inventario:
        incidencias_indice.append({"archivo": relativa, "motivo": "ruta duplicada en el índice", "detalle": ""})
        continue
    inventario[relativa] = {"ruta": ruta, "metadatos": fila}

if not inventario:
    raise ValueError(f"No hay PDF válidos en {indice}")

display(pd.DataFrame([f["metadatos"] for f in inventario.values()]).groupby("clasificadora").size().rename("PDF"))
if incidencias_indice:
    display(pd.DataFrame(incidencias_indice))
```

## 3. Inspeccionar una página
La vista y las coordenadas permiten localizar una tabla, imagen o gráfico. Los gráficos vectoriales se conservan en la vista de página. Los bloques y cuadros se leen en pasos posteriores.

```python
archivo_ejemplo = next(iter(inventario))
pagina_ejemplo = 1

with pymupdf.open(inventario[archivo_ejemplo]["ruta"]) as pdf:
    pagina = pdf[pagina_ejemplo - 1]
    display(Image(data=pagina.get_pixmap(dpi=110, alpha=False).tobytes("png")))
    print(f"{archivo_ejemplo} · página {pagina_ejemplo} de {len(pdf)}")
    display(pd.DataFrame([
        {"caja": list(imagen["bbox"]), "ancho_px": imagen["width"], "alto_px": imagen["height"]}
        for imagen in pagina.get_image_info()
    ]))
```

## 4. Leer texto, imágenes y tablas
El texto nativo se conserva por bloque. El OCR usa la página completa cuando hay menos de 50 caracteres de texto nativo, y regiones de imágenes grandes en páginas mixtas. El texto OCR queda separado para facilitar su revisión.

Las tablas nativas detectadas conservan filas y celdas. En una tabla escaneada indicada en `regiones_tabla_ocr`, se conservan las líneas y cajas de palabras recuperadas. La estructura de columnas en esas regiones se revisa con la imagen original.

```python
def caja_pdf(caja):
    return [round(coordenada, 2) for coordenada in caja]


def renderizar(pagina, region, dpi):
    visible = region * pagina.rotation_matrix & pagina.rect
    if visible.is_empty:
        raise ValueError("La región no intersecta la página")
    return pagina.get_pixmap(dpi=dpi, clip=visible, alpha=False)


def aplicar_ocr(pagina, region, psm=3):
    rotacion = pagina.rotation
    try:
        if rotacion:
            pagina.set_rotation(0)
        pix = renderizar(pagina, region, dpi_ocr)
    finally:
        if rotacion:
            pagina.set_rotation(rotacion)
    imagen = ImagenPIL.frombytes("RGB", (pix.width, pix.height), pix.samples)
    datos = pytesseract.image_to_data(
        imagen, lang="spa", config=f"--psm {psm}", timeout=90,
        output_type=pytesseract.Output.DICT,
    )
    lineas = {}
    escala = 72 / dpi_ocr
    for i, palabra in enumerate(datos["text"]):
        palabra = palabra.strip()
        if not palabra:
            continue
        x = (pix.x + datos["left"][i]) * escala
        y = (pix.y + datos["top"][i]) * escala
        caja = pymupdf.Rect(
            x, y, x + datos["width"][i] * escala, y + datos["height"][i] * escala,
        )
        clave = (datos["block_num"][i], datos["par_num"][i], datos["line_num"][i])
        lineas.setdefault(clave, []).append({
            "texto": palabra, "caja": caja_pdf(caja), "confianza": float(datos["conf"][i]),
        })
    filas = [
        {"texto": " ".join(p["texto"] for p in palabras), "palabras": palabras}
        for palabras in lineas.values()
    ]
    return {"caja": caja_pdf(region), "texto": "\n".join(f["texto"] for f in filas), "filas": filas}


def leer_pagina(pagina, relativa, carpeta_imagenes):
    numero = pagina.number + 1
    limites = pagina.rect * pagina.derotation_matrix
    bloques = [
        {"caja": caja_pdf(bloque[:4]), "texto": bloque[4].strip()}
        for bloque in pagina.get_text("blocks", sort=True) if bloque[6] == 0 and bloque[4].strip()
    ]
    texto_nativo = "\n".join(b["texto"] for b in bloques)
    registro = {
        "pagina": numero, "tamano": caja_pdf(limites), "rotacion": pagina.rotation,
        "texto_nativo": texto_nativo, "bloques": bloques, "ocr": [], "imagenes": [], "tablas": [],
        "incidencias": [],
    }
    if len(texto_nativo) < 50:
        try:
            registro["ocr"].append({"origen": "pagina", **aplicar_ocr(pagina, limites)})
        except (RuntimeError, ValueError) as error:
            registro["incidencias"].append({"etapa": "ocr_pagina", "detalle": str(error)})

    regiones_procesadas = set()
    for i, imagen in enumerate(pagina.get_image_info(), 1):
        region = pymupdf.Rect(imagen["bbox"]) & limites
        if region.is_empty:
            continue
        caja = caja_pdf(region)
        if tuple(caja) in regiones_procesadas:
            continue
        regiones_procesadas.add(tuple(caja))
        entrada = {"caja": caja, "ancho_px": imagen["width"], "alto_px": imagen["height"]}
        registro["imagenes"].append(entrada)
        try:
            if guardar_imagenes:
                ruta = carpeta_imagenes / f"p{numero:03d}-imagen-{i:02d}.png"
                ruta.parent.mkdir(parents=True, exist_ok=True)
                renderizar(pagina, region, 120).save(ruta)
                entrada["archivo"] = ruta.relative_to(salida).as_posix()
            if len(texto_nativo) >= 50 and region.get_area() / limites.get_area() >= area_minima_para_ocr:
                registro["ocr"].append({"origen": "imagen", **aplicar_ocr(pagina, region, psm=11)})
        except (RuntimeError, ValueError, OSError) as error:
            registro["incidencias"].append({"etapa": "imagen", "caja": caja, "detalle": str(error)})

    if len(texto_nativo) >= 50:
        try:
            for tabla in pagina.find_tables().tables:
                filas = tabla.extract()
                if any(celda and celda.strip() for fila in filas for celda in fila):
                    registro["tablas"].append({
                        "origen": "pdf", "caja": caja_pdf(tabla.bbox), "filas": filas,
                        "celdas": [[caja_pdf(celda) if celda else None for celda in fila.cells] for fila in tabla.rows],
                    })
        except (RuntimeError, ValueError) as error:
            registro["incidencias"].append({"etapa": "tablas_pdf", "detalle": str(error)})

    for coordenadas in regiones_tabla_ocr.get((relativa, numero), []):
        try:
            region = pymupdf.Rect(coordenadas) & limites
            registro["tablas"].append({"origen": "ocr", "requiere_revision": True, **aplicar_ocr(pagina, region, psm=6)})
        except (RuntimeError, ValueError) as error:
            registro["incidencias"].append({"etapa": "tabla_ocr", "detalle": str(error)})
    return registro
```

## 5. Extraer campos con evidencia
Se buscan valores rotulados de perspectiva y fecha del informe en el texto del PDF. Las coincidencias conservan su evidencia y los valores distintos pasan a revisión. La clasificación del catálogo sigue disponible en los metadatos.

```python
patrones = {
    "perspectiva": re.compile(
        r"(?im)^[ \t]*perspectiva(?:[ \t]+de[ \t]+la[ \t]+calificaci[o?]n)?[ \t]*(?:[:\-][ \t]*|\r?\n[ \t]*)(estable|positiva|negativa|en desarrollo)\b"
    ),
    "fecha_informe": re.compile(
        r"(?im)^\s*fecha\s+(?:del\s+informe|de\s+publicaci[oó]n)\s*[:\-]\s*([^\r\n]+)"
    ),
}


def extraer_campos(paginas):
    campos = {}
    for campo, patron in patrones.items():
        coincidencias = []
        for pagina in paginas:
            fuentes = [("pdf", pagina["texto_nativo"])]
            fuentes.extend(("ocr", fragmento["texto"]) for fragmento in pagina["ocr"])
            for origen, texto in fuentes:
                for hallazgo in patron.finditer(texto):
                    coincidencias.append({
                        "valor": hallazgo.group(1).strip(), "pagina": pagina["pagina"],
                        "origen": origen, "evidencia": hallazgo.group(0).strip(),
                    })
        valores = {hallazgo["valor"].casefold() for hallazgo in coincidencias}
        estado = "encontrado" if len(valores) == 1 else ("ambiguo" if valores else "sin_evidencia")
        campos[campo] = {
            "valor": coincidencias[0]["valor"] if estado == "encontrado" else None,
            "estado": estado, "evidencias": coincidencias,
        }
    return campos


def procesar(relativa, entrada):
    ruta = entrada["ruta"]
    resultado = {"archivo": relativa, "metadatos_indice": entrada["metadatos"], "paginas": [], "incidencias": []}
    try:
        with ruta.open("rb") as archivo:
            resultado["sha256"] = hashlib.file_digest(archivo, "sha256").hexdigest()
        id_ruta = hashlib.sha256(relativa.encode("utf-8")).hexdigest()[:16]
        carpeta_imagenes = salida / "imagenes" / id_ruta / resultado["sha256"][:16]
        with pymupdf.open(ruta) as pdf:
            if not pdf.is_pdf or pdf.needs_pass or not len(pdf):
                raise ValueError("PDF vacío, cifrado o inválido")
            for pagina in pdf:
                try:
                    resultado["paginas"].append(leer_pagina(pagina, relativa, carpeta_imagenes))
                except (RuntimeError, ValueError, OSError) as error:
                    resultado["incidencias"].append({"pagina": pagina.number + 1, "detalle": str(error)})
    except (RuntimeError, ValueError, OSError) as error:
        resultado["incidencias"].append({"etapa": "control", "detalle": str(error)})

    paginas = resultado["paginas"]
    resultado["campos"] = extraer_campos(paginas)
    incidencias = resultado["incidencias"] or any(p["incidencias"] for p in paginas)
    resultado["estado"] = "error" if not paginas else ("revisar" if incidencias else "ok")
    return resultado
```

## 7. Procesar y exportar
Cambia `lote_completo` a `True` para ejecutar los 48 PDF. La muestra y el lote guardan archivos separados. El JSONL contiene los resultados completos. El CSV resume la cobertura e indica los estados de los campos. Los recortes de im?genes se guardan como PNG y sus rutas son relativas a la carpeta de salida.

```python
lote_completo = False

muestra = {}
for relativa, entrada in inventario.items():
    muestra.setdefault(entrada["metadatos"]["clasificadora"], relativa)
seleccion = list(inventario) if lote_completo else list(muestra.values())

salida.mkdir(parents=True, exist_ok=True)
etiqueta = "informes" if lote_completo else "muestra"
destino = salida / f"{etiqueta}.jsonl"
resumen = []

with destino.open("w", encoding="utf-8") as archivo:
    for numero, relativa in enumerate(seleccion, 1):
        resultado = procesar(relativa, inventario[relativa])
        archivo.write(json.dumps(resultado, ensure_ascii=False, allow_nan=False) + "\n")
        paginas = resultado["paginas"]
        resumen.append({
            "archivo": relativa, "estado": resultado["estado"], "paginas": len(paginas),
            "paginas_con_texto": sum(bool(p["texto_nativo"] or any(o["texto"] for o in p["ocr"])) for p in paginas),
            "paginas_con_ocr": sum(bool(p["ocr"]) for p in paginas),
            "tablas": sum(len(p["tablas"]) for p in paginas),
            "imagenes": sum(len(p["imagenes"]) for p in paginas),
            "incidencias": len(resultado["incidencias"]) + sum(len(p["incidencias"]) for p in paginas),
            **{f"estado_{campo}": dato["estado"] for campo, dato in resultado["campos"].items()},
        })
        print(f"{numero}/{len(seleccion)} {relativa} · {resultado['estado']}")

tabla_resumen = pd.DataFrame(resumen)
tabla_resumen.to_csv(salida / f"{etiqueta}_resumen.csv", index=False, encoding="utf-8-sig")
pd.DataFrame(incidencias_indice, columns=["archivo", "motivo", "detalle"]).to_csv(
    salida / "incidencias_indice.csv", index=False, encoding="utf-8-sig",
)
display(tabla_resumen)
print(destino)
```

## 8. Comparativo de calidad

| Tarea | Motores | Indicadores |
|---|---|---|
| Texto | Tesseract, PaddleOCR, PyMuPDF, pdfplumber | CER, WER, tiempo y puntaje de OCR |
| Tablas | PyMuPDF, pdfplumber, tabula-py | Precisi?n, recall y F1 por celda y posici?n |
| Campos | Reglas, Qwen 3.5 4B y 9B en Ollama, y Qwen2.5 0.5B local en Hugging Face Transformers | F1 por campo, cobertura y valores sin evidencia |

Los casos se transcribieron visualmente de la p?gina 1 del informe de Agrobanco. Son una muestra de control, no una estimaci?n del rendimiento de todo el corpus. Ollama sirve los dos modelos Qwen 3.5. Los modelos y pesos opcionales se usan solo si est?n disponibles localmente. Los motores omitidos o con error conservan su motivo.

```python
from collections import Counter
from importlib import import_module
from time import perf_counter

from rapidfuzz.distance import Levenshtein
from mi_pipeline.calidad.f1 import f1_por_campo
from mi_pipeline.calidad.puntuar import puntuar

control = inventario["AGROBANCO/APOYO Y ASOCIADOS INTERNACIONALES/B+_e5c84757b53e.pdf"]["ruta"]
pag = 1
perspectiva = (409, 192, 475, 221)
indicadores = (411, 237, 518, 253)
comite = (409, 146, 583, 166)
activos = (414, 253, 572, 276)
tabla_area = (414, 252, 572, 334)
tabla_gold = [
    ["", "Jun-26", "Jun-25", "Dic-25"],
    ["Total Activos", "684", "590", "659"],
    ["Patrimonio", "608", "508", "585"],
    ["Resultado", "23", "31", "63"],
    ["ROA (Prom)", "6.9%", "10.9%", "10.4%"],
    ["ROE (Prom)", "7.8%", "12.7%", "11.9%"],
    ["Capital Global", "84.3%", "71.4%", "81.0%"],
]
casos_texto = {
    "perspectiva": (perspectiva, "Perspectiva Estable"),
    "encabezado": (indicadores, "Indicadores Financieros"),
    "comites": (comite, "Clasificaciones otorgadas en Comit?s de fecha 17/09/2026 y 18/03/2026"),
}
casos_campos = {
    "presentes": ([perspectiva, activos], {"perspectiva": "Estable", "activos_jun_2026": "684"}),
    "ausentes": ([indicadores], {"perspectiva": None, "activos_jun_2026": None}),
}
hf_modelo = "Qwen/Qwen2.5-0.5B-Instruct"
hf_cache = Path(os.getenv("HF_HOME", r"C:\Proyectos\modelos\hf")) / "hub"
paddle_cache = Path.home() / ".paddlex" / "official_models"
paddle_det = paddle_cache / "PP-OCRv5_mobile_det"
paddle_rec = paddle_cache / "latin_PP-OCRv5_mobile_rec"
resultados_benchmark = []


class NoDisponible(RuntimeError):
    pass


def importar(nombre):
    try:
        return import_module(nombre)
    except ImportError as error:
        raise NoDisponible(f"No se pudo importar {nombre}: {error}") from error


def ejecutar_motor(tarea, motor, casos, iniciar, medir):
    try:
        inicio = perf_counter()
        ejecutar = iniciar()
        carga = perf_counter() - inicio
    except Exception as error:
        estado = "omitido" if isinstance(error, NoDisponible) else "error"
        for nombre in casos:
            resultados_benchmark.append({
                "tarea": tarea, "motor": motor, "caso": nombre, "estado": estado,
                "detalle": f"{type(error).__name__}: {error}",
            })
        return
    for nombre, caso in casos.items():
        fila = {"tarea": tarea, "motor": motor, "caso": nombre, "estado": "ok", "carga_s": carga}
        inicio = perf_counter()
        try:
            predicho = ejecutar(caso)
            fila["segundos"] = perf_counter() - inicio
            fila.update(medir(caso, predicho))
        except Exception as error:
            fila.update(estado="omitido" if isinstance(error, NoDisponible) else "error",
                        detalle=f"{type(error).__name__}: {error}")
        resultados_benchmark.append(fila)


def norm(texto):
    return " ".join(str(texto).casefold().split())


def texto_pdf(caja):
    with pymupdf.open(control) as doc:
        return doc[pag - 1].get_text(clip=pymupdf.Rect(caja), sort=True)
```

```python
def tesseract_recorte(caso):
    with pymupdf.open(control) as doc:
        lectura = aplicar_ocr(doc[pag - 1], pymupdf.Rect(caso[0]), psm=6)
    conf = [p["confianza"] / 100 for fila in lectura["filas"] for p in fila["palabras"] if p["confianza"] >= 0]
    return lectura["texto"], sum(conf) / len(conf) if conf else 0.0


def iniciar_paddle():
    if not all(ruta.is_dir() for ruta in (paddle_det, paddle_rec)):
        raise NoDisponible(f"Configura las carpetas de pesos locales: {paddle_det} y {paddle_rec}")
    motor = importar("paddleocr").PaddleOCR(
        lang="es", ocr_version="PP-OCRv5", device="cpu",
        text_detection_model_dir=str(paddle_det), text_recognition_model_dir=str(paddle_rec),
        use_doc_orientation_classify=False, use_doc_unwarping=False, use_textline_orientation=False,
    )
    np = importar("numpy")

    def leer(caso):
        with pymupdf.open(control) as doc:
            pix = renderizar(doc[pag - 1], pymupdf.Rect(caso[0]), dpi_ocr)
        rgb = np.frombuffer(pix.samples, dtype=np.uint8).reshape(pix.height, pix.width, 3)
        pred = list(motor.predict(np.ascontiguousarray(rgb[:, :, ::-1])))
        conf = [float(s) for r in pred for s in r["rec_scores"]]
        return "\n".join(t for r in pred for t in r["rec_texts"]), sum(conf) / len(conf) if conf else 0.0
    return leer


def iniciar_plumber_texto():
    pdfplumber = importar("pdfplumber")
    def leer(caso):
        with pdfplumber.open(control) as doc:
            return doc.pages[pag - 1].crop(caso[0]).extract_text() or "", None
    return leer


def medir_texto(caso, lectura):
    texto, confianza = lectura
    a, b = norm(caso[1]), norm(texto)
    fila = {
        "cer": Levenshtein.distance(a, b) / len(a),
        "wer": Levenshtein.distance(a.split(), b.split()) / len(a.split()),
        "referencia": caso[1], "prediccion": texto,
    }
    if confianza is not None:
        calidad = puntuar(texto, confianza)
        fila.update(confianza=calidad["confianza"], validos=calidad["validos"], invalidos=calidad["invalidos"])
        fila["puntaje_ocr"] = calidad["score"] if calidad["validos"] + calidad["invalidos"] else None
    return fila


for motor, inicializar in {
    "Tesseract": lambda: tesseract_recorte,
    "PaddleOCR": iniciar_paddle,
    "PyMuPDF": lambda: lambda caso: (texto_pdf(caso[0]), None),
    "pdfplumber": iniciar_plumber_texto,
}.items():
    ejecutar_motor("texto", motor, casos_texto, inicializar, medir_texto)
```

```python
def pymupdf_tablas(caja):
    with pymupdf.open(control) as doc:
        return [t.extract() for t in doc[pag - 1].find_tables(clip=pymupdf.Rect(caja), strategy="text").tables]


def iniciar_plumber_tablas():
    pdfplumber = importar("pdfplumber")
    def leer(caja):
        with pdfplumber.open(control) as doc:
            return doc.pages[pag - 1].crop(caja).extract_tables({
                "vertical_strategy": "text", "horizontal_strategy": "text",
            })
    return leer


def iniciar_tabula():
    tabula = importar("tabula.io")
    java = shutil.which("java")
    if not java:
        javas = sorted((Path(os.getenv("ProgramFiles", r"C:\Program Files")) / "Eclipse Adoptium").glob("*/bin/java.exe"))
        if not javas:
            raise NoDisponible("Tabula necesita Java.")
        java = str(javas[-1])

    def leer(caja):
        x0, y0, x1, y1 = caja
        anterior = os.environ.get("PATH", "")
        try:
            os.environ["PATH"] = str(Path(java).parent) + os.pathsep + anterior
            tablas = tabula.read_pdf(
                str(control), pages=pag, area=[y0, x0, y1, x1], stream=True, guess=False,
                pandas_options={"header": None, "dtype": str},
                multiple_tables=True, force_subprocess=True,
            )
        finally:
            os.environ["PATH"] = anterior
        return [t.fillna("").values.tolist() for t in tablas]
    return leer


def medir_tabla(_, tablas):
    def celdas(filas, tabla):
        filas = [[norm(v) if v is not None else "" for v in f] for f in filas]
        return {(tabla, i, j, v) for i, f in enumerate(f for f in filas if any(f))
                for j, v in enumerate(f) if v}

    esperadas = celdas(tabla_gold, 0)
    obtenidas = set().union(*(celdas(filas, i) for i, filas in enumerate(tablas)))
    aciertos = len(esperadas & obtenidas)
    precision = aciertos / len(obtenidas) if obtenidas else 0.0
    recall = aciertos / len(esperadas)
    return {
        "precision": precision, "recall": recall,
        "f1": 2 * precision * recall / (precision + recall) if precision + recall else 0.0,
        "celdas_correctas": aciertos, "celdas_referencia": len(esperadas), "celdas_predichas": len(obtenidas),
        "referencia": tabla_gold, "prediccion": tablas,
    }


for motor, inicializar in {
    "PyMuPDF": lambda: pymupdf_tablas,
    "pdfplumber": iniciar_plumber_tablas,
    "tabula-py": iniciar_tabula,
}.items():
    ejecutar_motor("tablas", motor, {"indicadores_agrobanco": tabla_area}, inicializar, medir_tabla)
```

```python
contextos = {nombre: ("\n".join(texto_pdf(caja) for caja in cajas), gold)
             for nombre, (cajas, gold) in casos_campos.items()}
campos_prueba = list(next(iter(casos_campos.values()))[1])
esquema_prueba = {
    "type": "object", "additionalProperties": False, "required": campos_prueba,
    "properties": {campo: {
        "type": "object", "additionalProperties": False, "required": ["valor", "evidencia"],
        "properties": {"valor": {"type": ["string", "null"]}, "evidencia": {"type": ["string", "null"]}},
    } for campo in campos_prueba},
}


def mensaje(caso):
    return (
        "Extrae perspectiva y activos_jun_2026 en JSON. Los activos corresponden a la fila Total Activos "
        "y a la columna Jun-26. Para cada campo devuelve un objeto con valor como cadena y evidencia "
        "literal del documento. Si no aparece, devuelve valor y evidencia null. "
        f"Esquema: {json.dumps(esquema_prueba)}\nTexto:\n{caso[0]}"
    )


def iniciar_ollama(modelo):
    ollama = importar("ollama")
    cliente = ollama.Client(host="http://127.0.0.1:11434", timeout=120)
    try:
        disponibles = {m.model for m in ollama.Client(host="http://127.0.0.1:11434", timeout=4).list().models}
    except (OSError, RuntimeError, ollama.ResponseError) as error:
        raise NoDisponible(f"Servidor Ollama local no disponible: {error}") from error
    if modelo not in disponibles:
        raise NoDisponible(f"El servidor local no tiene {modelo}")
    def leer(caso):
        respuesta = cliente.chat(
            model=modelo, messages=[{"role": "user", "content": mensaje(caso)}],
            format=esquema_prueba, think=False, options={"temperature": 0, "num_predict": 450},
        )
        return json.loads(respuesta.message.content)
    return leer


def iniciar_hf():
    hf = importar("transformers")
    try:
        tokenizer = hf.AutoTokenizer.from_pretrained(hf_modelo, cache_dir=hf_cache, local_files_only=True)
        modelo = hf.AutoModelForCausalLM.from_pretrained(hf_modelo, cache_dir=hf_cache, local_files_only=True)
    except OSError as error:
        raise NoDisponible(f"{hf_modelo} no se pudo cargar desde la cach? local: {error}") from error
    modelo.eval()
    torch = importar("torch")

    def leer(caso):
        prompt = tokenizer.apply_chat_template(
            [{"role": "user", "content": mensaje(caso)}], tokenize=False, add_generation_prompt=True,
        )
        entrada = tokenizer(prompt, return_tensors="pt")
        with torch.inference_mode():
            salida_modelo = modelo.generate(**entrada, max_new_tokens=320, do_sample=False)
        texto = tokenizer.decode(salida_modelo[0][entrada.input_ids.shape[1]:], skip_special_tokens=True).strip()
        objeto = re.search(r"\{.*\}", texto, re.S)
        if not objeto:
            raise ValueError("El modelo no devolvi? un objeto JSON")
        return json.loads(objeto.group())
    return leer


def reglas_control(caso):
    texto = caso[0]
    patrones_control = {
        "perspectiva": r"(?i)\bPerspectiva\s+(Estable|Positiva|Negativa)\b",
        "activos_jun_2026": r"(?is)\bJun-26\b.*?\bTotal Activos\s+([\d,]+)",
    }
    resultado = {}
    for campo, patron in patrones_control.items():
        match = re.search(patron, texto)
        resultado[campo] = {"valor": match[1], "evidencia": match[0]} if match else {"valor": None, "evidencia": None}
    return resultado


def medir_campos(caso, respuesta):
    texto, gold = caso
    if not isinstance(respuesta, dict):
        raise ValueError("La predicci?n no contiene un objeto de campos.")
    predichos = {}
    sin_evidencia = []
    for campo, dato in respuesta.items():
        if campo not in campos_prueba or not isinstance(dato, dict):
            raise ValueError(f"Campo inesperado o mal formado: {campo}")
        valor, evidencia = dato.get("valor"), dato.get("evidencia")
        if valor is None or not str(valor).strip():
            continue
        predichos[campo] = str(valor).strip()
        literal = norm(evidencia or "")
        valor_literal = re.search(r"(?<!\w)" + re.escape(norm(valor)) + r"(?!\w)", literal)
        if not literal or literal not in norm(texto) or not valor_literal:
            sin_evidencia.append(campo)
    return {
        "referencia": gold, "prediccion": predichos, "respuesta": respuesta,
        "valores_emitidos": len(predichos), "valores_sin_evidencia": len(sin_evidencia),
        "campos_sin_evidencia": sin_evidencia,
    }


for motor, inicializar in {
    "Reglas": lambda: reglas_control,
    "Ollama / qwen3.5:4b": lambda: iniciar_ollama("qwen3.5:4b"),
    "Ollama / qwen3.5:9b": lambda: iniciar_ollama("qwen3.5:9b"),
    f"Hugging Face / {hf_modelo}": iniciar_hf,
}.items():
    ejecutar_motor("campos", motor, contextos, inicializar, medir_campos)
```

### Interpretaci?n y exportaci?n
CER y WER miden errores de caracteres y palabras respecto de la transcripci?n. El puntaje OCR combina la confianza declarada con validadores de RUC, fechas y montos. Se deja vac?o cuando la regi?n carece de esos valores. El F1 de tablas exige coincidencia de valor y posici?n. El F1 de campos compara con la referencia humana. ?Sin evidencia? se?ala una posible alucinaci?n por falta de respaldo literal. Un valor con evidencia puede pertenecer a otra columna y aparecer como error de F1.

```python
detalle = pd.DataFrame(resultados_benchmark)
indicadores_motor = []
indicadores_campo = []
for (tarea, motor), filas in detalle.groupby(["tarea", "motor"], sort=False):
    exitos = filas[filas["estado"].eq("ok")]
    resumen_motor = {
        "tarea": tarea, "motor": motor, "casos_ok": len(exitos),
        "omitidos": int(filas["estado"].eq("omitido").sum()),
        "errores": int(filas["estado"].eq("error").sum()),
    }
    for metrica in ("cer", "wer", "puntaje_ocr", "confianza", "precision", "recall", "f1", "segundos"):
        if metrica in exitos and exitos[metrica].notna().any():
            resumen_motor[metrica] = exitos[metrica].mean()
    if tarea == "campos" and len(exitos):
        aciertos = f1_por_campo(exitos["prediccion"].tolist(), exitos["referencia"].tolist(), campos_prueba)
        indicadores_campo.extend({"motor": motor, "campo": k, **v} for k, v in aciertos.items())
        resumen_motor["f1"] = sum(v["f1"] for v in aciertos.values()) / len(aciertos)
        emitidos = int(exitos["valores_emitidos"].sum())
        esperados = [{k for k, v in ref.items() if v is not None} for ref in exitos["referencia"]]
        total_esperados = sum(map(len, esperados))
        presentes = sum(len(e & pred.keys()) for e, pred in zip(esperados, exitos["prediccion"]))
        resumen_motor["cobertura"] = presentes / total_esperados if total_esperados else None
        resumen_motor["tasa_sin_evidencia"] = exitos["valores_sin_evidencia"].sum() / emitidos if emitidos else None
    indicadores_motor.append(resumen_motor)

resumen_benchmark = pd.DataFrame(indicadores_motor)
por_campo = pd.DataFrame(indicadores_campo)
display(resumen_benchmark.round(3))
display(por_campo)
fallos = detalle[detalle["estado"].ne("ok")]
if not fallos.empty:
    display(fallos[["tarea", "motor", "estado", "detalle"]])

destino_benchmark = salida / "comparativo"
destino_benchmark.mkdir(parents=True, exist_ok=True)
resumen_benchmark.to_csv(destino_benchmark / "resumen.csv", index=False, encoding="utf-8-sig")
por_campo.to_csv(destino_benchmark / "f1_por_campo.csv", index=False, encoding="utf-8-sig")
detalle.to_json(destino_benchmark / "detalle.jsonl", orient="records", lines=True, force_ascii=False)
```
