# Extracción de informes bancarios
Ejecuta las celdas en orden. El índice aporta los metadatos del catálogo. El texto y las tablas se leen del PDF. El primer recorrido procesa un informe por clasificadora. Puedes activar el lote completo en la última sección.

```python
import csv
from pathlib import Path

import pymupdf
import pandas as pd
from IPython.display import Image, display

from mi_pipeline.orquestacion.config import cargar_config
from mi_pipeline.orquestacion.trabajo import procesar_archivo

raiz = Path.cwd()
carpeta_pdf = (raiz / "datos" / "Banco").resolve()
indice = carpeta_pdf / "indice.csv"
with indice.open(encoding="utf-8-sig", newline="") as archivo:
    catalogo = list(csv.DictReader(archivo))

informes = []
rutas_invalidas = []
for fila in catalogo:
    if fila["estado"] != "descargado":
        continue
    ruta = (carpeta_pdf / fila["archivo"].replace(chr(92), "/")).resolve()
    if not ruta.is_relative_to(carpeta_pdf) or ruta.suffix.lower() != ".pdf" or not ruta.is_file():
        rutas_invalidas.append(fila["archivo"])
        continue
    informes.append((fila, ruta))

if rutas_invalidas:
    raise ValueError(f"PDF faltantes o inválidos en el índice: {rutas_invalidas}")
if not informes:
    raise ValueError(f"El índice no contiene PDF disponibles: {indice}")

config = cargar_config()
print(f"{len(informes)} PDF disponibles")
display(pd.DataFrame([fila for fila, _ in informes]).groupby("clasificadora").size().rename("informes"))
```

## Inspeccionar un informe
Cambia la posición del informe y el número de página para explorar el contenido. La vista muestra la página completa, los bloques de texto, el número de imágenes y las tablas nativas detectadas en esa página.

```python
posicion = 0
numero_pagina = 1
metadatos, ruta_muestra = informes[posicion]

with pymupdf.open(ruta_muestra) as pdf:
    pagina = pdf[numero_pagina - 1]
    display(Image(data=pagina.get_pixmap(dpi=110).tobytes("png")))
    print(f"Imágenes: {len(pagina.get_image_info())}")
    print(pagina.get_text()[:1500])
    tablas_pagina = [tabla.extract() for tabla in pagina.find_tables().tables]

for filas in tablas_pagina:
    display(pd.DataFrame(filas))
```

## Extraer texto y examinar el resultado
El pipeline valida el PDF, lee el texto nativo y aplica OCR a páginas con poca capa de texto. La calificación del índice se conserva como metadato. Los informes bancarios todavía carecen de un esquema de campos específico y las tablas escaneadas requieren revisión visual.

```python
tipo = "informe_clasificacion_bancaria"
resultado_muestra = procesar_archivo(ruta_muestra, tipo, config)
print("Estado:", resultado_muestra["estado"])
print("Páginas enviadas a OCR:", resultado_muestra.get("paginas_ocr", 0))
print(resultado_muestra.get("detalle") or resultado_muestra.get("texto", "")[:2200])
```

## Procesar una muestra o el lote completo
Por defecto se selecciona un PDF por clasificadora. Cambia `lote_completo` a `True` para procesar todos los informes. La exportación guarda metadatos, texto, tablas nativas con su página y errores por documento en un JSONL local.

```python
lote_completo = False
muestra_por_clasificadora = {}
for fila, ruta in informes:
    muestra_por_clasificadora.setdefault(fila["clasificadora"], (fila, ruta))

seleccion = informes if lote_completo else list(muestra_por_clasificadora.values())
resultados = []
for fila, ruta in seleccion:
    resultado = procesar_archivo(ruta, tipo, config)
    tablas = []
    errores_tablas = []
    if resultado["estado"] == "ok":
        with pymupdf.open(ruta) as pdf:
            for pagina in pdf:
                try:
                    tablas.extend(
                        {"pagina": pagina.number + 1, "filas": tabla.extract()}
                        for tabla in pagina.find_tables().tables
                    )
                except (RuntimeError, ValueError) as error:
                    errores_tablas.append({"pagina": pagina.number + 1, "detalle": str(error)})
    resultados.append({
        "metadatos": fila,
        **resultado,
        "tablas_nativas": tablas,
        "errores_tablas": errores_tablas,
    })
    print(f"{len(resultados)}/{len(seleccion)} {fila['archivo']}: {resultado['estado']}")

display(pd.DataFrame([{
    "archivo": r["metadatos"]["archivo"],
    "estado": r["estado"],
    "caracteres": r.get("caracteres", 0),
    "tablas_nativas": len(r["tablas_nativas"]),
} for r in resultados]))
```

```python
import json

salida = raiz / "salidas" / "banco"
salida.mkdir(parents=True, exist_ok=True)
nombre = "informes.jsonl" if lote_completo else "muestra.jsonl"
destino = salida / nombre
with destino.open("w", encoding="utf-8") as archivo:
    for resultado in resultados:
        archivo.write(json.dumps(resultado, ensure_ascii=False) + "\n")
print(destino)
```
